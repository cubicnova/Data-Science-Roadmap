# 09 – File Handling in Python

The notebook creates a dedicated temporary practice folder so examples do not overwrite your project files.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

workspace = TemporaryDirectory()
root = Path(workspace.name)
print("Practice directory:", root)
```

## Write, read, and append text

```python
text_path = root / "notes.txt"
with text_path.open("w", encoding="utf-8") as file:
    file.write("First line\nSecond line\n")
with text_path.open("a", encoding="utf-8") as file:
    file.write("Third line\n")
print(text_path.read_text(encoding="utf-8"))
```

## File paths and metadata

```python
print("Name:", text_path.name)
print("Suffix:", text_path.suffix)
print("Exists:", text_path.exists())
print("Is file:", text_path.is_file())
print("Size:", text_path.stat().st_size, "bytes")
```

## Process a file line by line

```python
line_count = 0
with text_path.open(encoding="utf-8") as file:
    for line in file:
        line_count += 1
        print(f"{line_count}: {line.rstrip(chr(10))}")
print("Line count:", line_count)
```

## CSV files

```python
import csv
csv_path = root / "scores.csv"
rows = [{"name": "Asha", "score": 91}, {"name": "Ravi", "score": 84}]
with csv_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)
with csv_path.open(newline="", encoding="utf-8") as file:
    loaded_rows = list(csv.DictReader(file))
print(loaded_rows)
```

## JSON files

```python
import json
json_path = root / "config.json"
config = {"project": "data-science-roadmap", "active": True, "version": 1}
with json_path.open("w", encoding="utf-8") as file:
    json.dump(config, file, indent=2)
with json_path.open(encoding="utf-8") as file:
    loaded_config = json.load(file)
print(loaded_config)
```

## Handling missing files

```python
missing_path = root / "missing.txt"
try:
    missing_path.read_text(encoding="utf-8")
except FileNotFoundError:
    print("File was not found; handle this case explicitly.")
```

## Binary copying

```python
source_bin = root / "sample.bin"
copy_bin = root / "sample-copy.bin"
source_bin.write_bytes(bytes([0, 1, 2, 255]))
with source_bin.open("rb") as source, copy_bin.open("wb") as destination:
    destination.write(source.read())
print(copy_bin.read_bytes())
```
