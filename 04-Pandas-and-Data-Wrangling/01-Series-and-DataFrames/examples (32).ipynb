# Pandas Series and DataFrames

Runnable introduction to Pandas' two core data structures.

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', 10)
```

## 1. Create a Series

```python
scores = pd.Series([78, 85, 92], index=['Asha', 'Dev', 'Mira'], name='score')
print(scores)
print('Dev:', scores.loc['Dev'], '| second position:', scores.iloc[1])
```

## 2. Create and inspect a DataFrame

```python
df = pd.DataFrame({'name': ['Asha', 'Dev', 'Mira'], 'age': [21, 22, 20], 'score': [78, 85, 92]})
print(df)
print('shape:', df.shape)
print('columns:', df.columns.tolist())
print('dtypes:\n', df.dtypes)
```

## 3. Select rows and columns

```python
print('one column:\n', df['score'])
print('two columns:\n', df[['name', 'score']])
print('label selection:', df.loc[1, 'name'])
print('position selection:', df.iloc[1, 2])
```

## 4. Filter, add, rename, and drop

```python
df['passed'] = df['score'] >= 40
print(df[df['score'] >= 80])
renamed = df.rename(columns={'score': 'marks'})
print(renamed.drop(columns=['passed']))
```

## 5. Index management

```python
indexed = df.set_index('name')
print(indexed)
print(indexed.reset_index())
```

## 6. Summary statistics

```python
print(df.describe())
print('average score:', df['score'].mean())
print('score counts:', df['score'].value_counts())
```

## 7. Missing values

```python
messy = pd.DataFrame({'employee': ['A', 'B', 'C'], 'salary': [50000.0, np.nan, 65000.0]})
print('missing counts:\n', messy.isna().sum())
messy['salary'] = messy['salary'].fillna(messy['salary'].median())
print(messy)
```

## 8. Series alignment

```python
a = pd.Series([10, 20], index=['x', 'y'])
b = pd.Series([1, 2], index=['y', 'x'])
print(a + b)
```

## 9. Sort and convert data types

```python
data = pd.DataFrame({'name': ['Asha', 'Dev', 'Mira'], 'score': ['78', '92', '85']})
data['score'] = pd.to_numeric(data['score'])
print(data.sort_values('score', ascending=False))
print(data.dtypes)
```

## 10. Mini-project: student marks

```python
marks = pd.DataFrame({'name': ['Asha', 'Dev', 'Mira'], 'math': [80, 70, 90], 'science': [85, 75, 95]})
marks['total'] = marks[['math', 'science']].sum(axis=1)
marks['average'] = marks[['math', 'science']].mean(axis=1)
print(marks)
assert marks.shape == (3, 5)
```
