## 1. Setup and imports

Install any missing dependencies in your environment with `python -m pip install numpy pandas matplotlib scikit-learn jupyter`.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris, load_diabetes, make_blobs
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.dummy import DummyClassifier, DummyRegressor
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    mean_absolute_error, mean_squared_error, r2_score
)

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
```

## 2. What kind of ML problem is it?

- **Classification:** predict a category.
- **Regression:** predict a numeric value.
- **Clustering:** discover groups without target labels.
- **Reinforcement learning:** learn actions from rewards and feedback.

## 3. Load and inspect the Iris dataset

The Iris dataset contains flower measurements and species labels. The four measurements are features; the species is the target.

```python
iris = load_iris()
X = iris.data
y = iris.target

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("Features:", iris.feature_names)
print("Classes:", iris.target_names)
print("Class counts:", np.bincount(y))

iris_df = pd.DataFrame(X, columns=iris.feature_names)
iris_df["species"] = [iris.target_names[i] for i in y]
iris_df.head()
```

## 4. Split the data

The test set is held out to estimate performance on unseen observations. `stratify=y` preserves class proportions across the split.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Training class counts:", np.bincount(y_train))
print("Testing class counts:", np.bincount(y_test))
```

## 5. Train a simple classifier

Scikit-learn estimators generally use `fit()` to learn from training data and `predict()` to produce predictions.

```python
classifier = LogisticRegression(max_iter=1000, random_state=42)
classifier.fit(X_train, y_train)

y_pred = classifier.predict(X_test)
print("First 10 predictions:", y_pred[:10])
print("First 10 actual labels:", y_test[:10])
```

## 6. Evaluate classification performance

Accuracy is the fraction of correct predictions. The classification report includes precision, recall, and F1-score by class.

```python
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.3f}")
print("\nConfusion matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification report:\n")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
```

## 7. Compare against a simple baseline

A `DummyClassifier` ignores the feature relationships and predicts the most frequent training class. A useful model should be evaluated against a simple baseline.

```python
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
baseline_pred = baseline.predict(X_test)

print(f"Dummy baseline accuracy: {accuracy_score(y_test, baseline_pred):.3f}")
print(f"Logistic Regression accuracy: {accuracy_score(y_test, y_pred):.3f}")
```

## 8. Regression example with the diabetes dataset

Regression predicts a numeric target. We will fit `LinearRegression` and measure MAE and R².

```python
diabetes = load_diabetes()
X_reg, y_reg = diabetes.data, diabetes.target

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.20, random_state=42
)

regressor = LinearRegression()
regressor.fit(X_train_reg, y_train_reg)
y_reg_pred = regressor.predict(X_test_reg)

print(f"MAE: {mean_absolute_error(y_test_reg, y_reg_pred):.3f}")
print(f"R²: {r2_score(y_test_reg, y_reg_pred):.3f}")
```

## 9. Compare regression with a mean baseline

```python
regression_baseline = DummyRegressor(strategy="mean")
regression_baseline.fit(X_train_reg, y_train_reg)
baseline_reg_pred = regression_baseline.predict(X_test_reg)

for name, predictions in [
    ("Mean baseline", baseline_reg_pred),
    ("Linear Regression", y_reg_pred),
]:
    print(name)
    print("  MAE:", round(mean_absolute_error(y_test_reg, predictions), 3))
    print("  R²:", round(r2_score(y_test_reg, predictions), 3))
```

## 10. Visualize actual vs. predicted values

Points closer to the diagonal line indicate predictions closer to actual values. This visualization is only a diagnostic, not a complete evaluation.

```python
plt.figure(figsize=(6, 5))
plt.scatter(y_test_reg, y_reg_pred, alpha=0.75)
low = min(y_test_reg.min(), y_reg_pred.min())
high = max(y_test_reg.max(), y_reg_pred.max())
plt.plot([low, high], [low, high], linestyle="--")
plt.xlabel("Actual target")
plt.ylabel("Predicted target")
plt.title("Diabetes dataset: actual vs. predicted")
plt.tight_layout()
plt.show()
```

## 11. Key takeaways

- `X` contains features; `y` contains the target for supervised learning.
- Split data before fitting the model.
- `fit()` trains; `predict()` generates outputs.
- Classification and regression need different metrics.
- Baselines help establish whether a model adds value.
- High training performance does not guarantee good performance on unseen data.
- In production workflows, prevent leakage, check data quality and fairness, and monitor model behavior.
