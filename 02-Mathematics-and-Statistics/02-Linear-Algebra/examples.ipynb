# Linear Algebra for Data Science

This notebook uses NumPy to demonstrate vector and matrix operations.

## 1. Vectors and shapes

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
A = np.array([[1, 2, 3], [4, 5, 6]])
print('a:', a, 'shape:', a.shape)
print('A shape:', A.shape)
```

## 2. Vector operations and dot product

```python
print('sum:', a + b)
print('scalar product:', 3 * a)
print('dot product:', a @ b)
```

## 3. Matrix operations

```python
M = np.array([[1, 2], [3, 4]])
N = np.array([[5, 6], [7, 8]])
print('M + N:\n', M + N)
print('M @ N:\n', M @ N)
print('M transpose:\n', M.T)
```

## 4. Matrix-vector multiplication

```python
features = np.array([[1, 2], [3, 4]])
weights = np.array([5, 6])
print(features @ weights)
```

## 5. Norms and distances

```python
x = np.array([3, 4])
y = np.array([6, 8])
print('norm of x:', np.linalg.norm(x))
print('distance:', np.linalg.norm(x - y))
print('L1 norm:', np.abs(np.array([-2, 3, -5])).sum())
```

## 6. Solving a linear system

```python
coefficients = np.array([[2, 1], [1, -1]])
constants = np.array([5, 1])
solution = np.linalg.solve(coefficients, constants)
print('solution [x, y]:', solution)
print('verification:', coefficients @ solution)
```
