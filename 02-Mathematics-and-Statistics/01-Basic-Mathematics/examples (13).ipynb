# Basic Mathematics for Data Science

A worked notebook for arithmetic, percentages, algebra, powers, averages, and Python calculations.

## 1. Order of operations

```python
a = 18 + 6 * 3
b = (18 + 6) * 3
print(a, b)
```

## 2. Fractions and percentages

```python
fraction = 3 / 5
percentage = 22
amount = 450
print('fraction as decimal:', fraction)
print('fraction as percent:', fraction * 100)
print('22% of 450:', amount * percentage / 100)
```

## 3. Percentage change

```python
old_value, new_value = 800, 920
change = (new_value - old_value) / old_value * 100
print(f'Percentage change: {change:.1f}%')
```

## 4. Algebra

```python
x = (37 - 9) / 4  # solves 4x + 9 = 37
print('x =', x)
print('f(3) for f(x)=4x-2:', 4 * 3 - 2)
```

## 5. Exponents, roots, and logarithms

```python
import math
print('2**6 =', 2 ** 6)
print('sqrt(196) =', math.sqrt(196))
print('log10(10000) =', math.log10(10000))
print('log2(32) =', math.log2(32))
```

## 6. Mean and absolute differences

```python
values = [4, 6, 6, 8, 11]
mean = sum(values) / len(values)
absolute_differences = [abs(value - mean) for value in values]
print('mean:', mean)
print('absolute differences:', absolute_differences)
```

## 7. Safe percentage change helper

```python
def percentage_change(old_value, new_value):
    if old_value == 0:
        raise ValueError('Percentage change is undefined when old_value is zero.')
    return (new_value - old_value) / old_value * 100

print(percentage_change(250, 200))
```
