# NumPy Array Operations — Runnable Examples

This notebook demonstrates vectorized arithmetic, ufuncs, comparisons, aggregations, broadcasting, reshaping, sorting, combining arrays, and linear algebra.

```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 1. Vectorized arithmetic

```python
a = np.array([2, 4, 6])
b = np.array([1, 3, 5])
print("add:", a + b)
print("subtract:", a - b)
print("multiply element-wise:", a * b)
print("divide:", a / b)
print("square:", a ** 2)
```

## 2. Universal functions

```python
x = np.array([1.0, 4.0, 9.0, 16.0])
print("square roots:", np.sqrt(x))
print("absolute:", np.abs(np.array([-4, -1, 0, 3])))
angles = np.array([0, np.pi / 2, np.pi])
print("sine:", np.sin(angles))
print("cosine:", np.cos(angles))
print("exponential:", np.exp(np.array([0.0, 1.0])))
print("natural log:", np.log(np.array([1.0, np.e])))
```

## 3. Comparisons and filtering

```python
values = np.array([3, 8, 12, 5, 17, 22])
print("greater than 10:", values[values > 10])
print("between 5 and 15:", values[(values >= 5) & (values <= 15)])
print("replace values below 10 with zero:", np.where(values < 10, 0, values))
```

## 4. Aggregations and axis

```python
m = np.array([[1, 2, 3], [4, 5, 6]])
print("sum:", m.sum())
print("mean:", m.mean())
print("column sums:", m.sum(axis=0))
print("row sums:", m.sum(axis=1))
print("row means keeping dimensions:\n", m.mean(axis=1, keepdims=True))
print("population std:", m.std(ddof=0))
print("sample std:", m.std(ddof=1))
print("argmax flat index:", m.argmax())
```

## 5. Broadcasting

```python
column = np.array([[1], [2], [3]])
row = np.array([[10, 20, 30, 40]])
result = column + row
print(result)
print("result shape:", result.shape)
```

## 6. Reshape, transpose, and flatten

```python
a = np.arange(12)
m = a.reshape(3, 4)
print("matrix:\n", m)
print("transpose:\n", m.T)
print("ravel:", m.ravel())
print("flatten:", m.flatten())
```

## 7. Sorting, unique values, and searching

```python
a = np.array([4, 1, 4, 2, 1])
print("sorted:", np.sort(a))
print("descending:", np.sort(a)[::-1])
print("unique:", np.unique(a))
print("argsort:", np.argsort(a))
print("indices greater than 2:", np.where(a > 2))
```

## 8. Combining and splitting

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("concatenate:", np.concatenate([a, b]))
print("stack:\n", np.stack([a, b], axis=0))
print("vertical stack:\n", np.vstack([a, b]))
print("split:", np.split(np.arange(8), 4))
```

## 9. Matrix operations

```python
A = np.array([[2, 1], [1, 3]], dtype=float)
B = np.array([[1, 2], [3, 4]], dtype=float)
print("A @ B:\n", A @ B)
print("determinant of A:", np.linalg.det(A))
print("solution to A x = [5, 6]:", np.linalg.solve(A, np.array([5, 6], dtype=float)))
```

## 10. Mini task: adjust marks and summarize

```python
names = np.array(["Asha", "Ravi", "Meena", "Dev"])
marks = np.array([68, 74, 92, 99])
adjusted = np.minimum(marks + 5, 100)
print("adjusted marks:", adjusted)
print("mean:", adjusted.mean())
print("students scoring at least 75:", names[adjusted >= 75])
assert np.all(adjusted <= 100)
assert np.array_equal(adjusted, np.array([73, 79, 97, 100]))
print("All checks passed.")
```
