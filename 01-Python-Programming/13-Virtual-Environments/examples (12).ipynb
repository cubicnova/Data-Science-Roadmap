# Python Virtual Environments

This notebook demonstrates how to inspect the current Python interpreter, inspect package metadata, and understand dependency files. Environment creation and activation commands are shell-specific and are documented as text rather than executed in this notebook.

## 1. Inspect the active interpreter

````python
import sys
print('Python version:', sys.version.split()[0])
print('Interpreter:', sys.executable)
````

## 2. Inspect pip

````python
import subprocess, sys
result = subprocess.run([sys.executable, '-m', 'pip', '--version'], capture_output=True, text=True)
print(result.stdout.strip() or result.stderr.strip())
````

## 3. Create an environment from a terminal

Windows: `py -m venv .venv`

macOS/Linux: `python3 -m venv .venv`

Activate with the command for your shell, then use `python -m pip` to install dependencies.

## 4. Example dependency file

A `requirements.txt` file can contain one package requirement per line, for example `requests>=2.31,<3`. Review version constraints for your project before using them.

````python
example_requirements = ['requests>=2.31,<3', 'urllib3>=2,<3']
print('\n'.join(example_requirements))
````

## 5. Check an installed package without requiring it to be present

````python
from importlib import metadata
package_name = 'requests'
try:
    print(package_name, metadata.version(package_name))
except metadata.PackageNotFoundError:
    print(f'{package_name} is not installed in this interpreter.')
````

## 6. Example workflow commands

```bash
python3 -m venv .venv
source .venv/bin/activate
python -m pip install -r requirements.txt
python -m pip freeze > requirements.txt
deactivate
```

On Windows, use the PowerShell or Command Prompt activation command listed in `README.md`.
