# 10 – Exception Handling in Python

Examples demonstrate controlled error handling. Expected exceptions are caught so the notebook can continue.

```python
try:
    value = int("42")
except ValueError:
    print("Invalid integer")
else:
    print("Parsed:", value)
```

## Handle multiple expected errors

```python
def divide_text(a_text, b_text):
    try:
        a = float(a_text)
        b = float(b_text)
        return a / b
    except ValueError:
        return "Both inputs must be numbers."
    except ZeroDivisionError:
        return "The denominator must not be zero."

print(divide_text("12", "3"))
print(divide_text("12", "0"))
print(divide_text("twelve", "3"))
```

## Raise exceptions for invalid input

```python
def percentage(part, whole):
    if whole == 0:
        raise ZeroDivisionError("whole must not be zero")
    if part < 0 or whole < 0:
        raise ValueError("values must be non-negative")
    return part / whole * 100

print(percentage(25, 100))
try:
    percentage(5, 0)
except ZeroDivisionError as exc:
    print(type(exc).__name__ + ":", exc)
```

## Custom exceptions

```python
class InsufficientFundsError(Exception):
    pass

def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError("amount must be positive")
    if amount > balance:
        raise InsufficientFundsError("balance is too low")
    return balance - amount

print(withdraw(100, 30))
try:
    withdraw(100, 150)
except InsufficientFundsError as exc:
    print(type(exc).__name__ + ":", exc)
```

## Exception chaining

```python
def parse_score(text):
    try:
        return int(text)
    except ValueError as exc:
        raise ValueError("score must be an integer") from exc

try:
    parse_score("high")
except ValueError as exc:
    print("User-facing error:", exc)
    print("Original cause:", type(exc.__cause__).__name__)
```

## Logging exceptions

```python
import logging
logger = logging.getLogger("exception_demo")
logger.setLevel(logging.ERROR)
if not logger.handlers:
    logger.addHandler(logging.StreamHandler())
try:
    int("not-a-number")
except ValueError:
    logger.exception("Could not parse the supplied integer")
```

## Cleanup with finally

```python
try:
    print("Performing a task")
finally:
    print("Cleanup step runs even if an exception occurs")
```
