# NumPy Array Indexing and Slicing — Runnable Examples

Run these cells from top to bottom to practice indexing, slicing, advanced indexing, Boolean masks, views/copies, and dimension changes.

```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 1. Positive and negative indices

```python
a = np.array([10, 20, 30, 40, 50])
print("first:", a[0])
print("third:", a[2])
print("last and second-last:", a[-1], a[-2])
```

## 2. Slice syntax: start, stop, step

```python
a = np.arange(10)
print("indices 2 through 6:", a[2:7])
print("first four:", a[:4])
print("from index 5:", a[5:])
print("every second:", a[::2])
print("reversed:", a[::-1])
```

## 3. 2D indexing

```python
m = np.arange(1, 13).reshape(3, 4)
print("matrix:\n", m)
print("row 1, col 2:", m[1, 2])
print("second row:", m[1, :])
print("first column:", m[:, 0])
print("last column:", m[:, -1])
print("top-left 2x2:\n", m[:2, :2])
```

## 4. Fancy indexing and Cartesian selection

```python
m = np.arange(9).reshape(3, 3)
print("paired coordinates:", m[[0, 2], [2, 1]])
print("Cartesian row/column selection:\n", m[np.ix_([0, 2], [1, 2])])
```

## 5. Views versus copies

```python
original = np.arange(5)
view = original[1:4]
view[0] = 99
print("original after view mutation:", original)
print("view shares memory:", np.shares_memory(original, view))
original2 = np.arange(5)
selected = original2[[1, 2, 3]]
selected[0] = 99
print("original after advanced-index mutation:", original2)
print("advanced selection shares memory:", np.shares_memory(original2, selected))
```

## 6. Boolean indexing

```python
values = np.array([3, 8, 12, 5, 17, 22])
print("greater than 10:", values[values > 10])
print("between 5 and 15 inclusive:", values[(values >= 5) & (values <= 15)])
data = np.array([[18, 70], [21, 82], [19, 65], [25, 91]])
print("rows with score >= 70:\n", data[data[:, 1] >= 70])
```

## 7. Add dimensions

```python
a = np.array([1, 2, 3])
column = a[:, np.newaxis]
row = a[np.newaxis, :]
print("original shape:", a.shape)
print("column shape:", column.shape)
print("row shape:", row.shape)
```

## 8. Practical mini task

```python
data = np.array([
    [18, 70, 1],
    [21, 82, 0],
    [19, 65, 1],
    [25, 91, 0],
])
selected_rows = data[data[:, 1] >= 70]
final = selected_rows[:, [0, 1]]
print(final)
print("final shape:", final.shape)
assert final.shape == (3, 2)
assert np.array_equal(final[:, 1], np.array([70, 82, 91]))
print("All checks passed.")
```
