# Probability for Data Science

Examples cover sample spaces, event rules, conditional probability, Bayes' theorem, expectation, and simulation.

## 1. Equally likely outcomes and complements

```python
sample_space = {1, 2, 3, 4, 5, 6}
even = {2, 4, 6}
p_even = len(even) / len(sample_space)
print('P(even):', p_even)
p_rain = 0.3
print('P(no rain):', 1 - p_rain)
```

## 2. Addition and multiplication rules

```python
p_a, p_b, p_intersection = 0.4, 0.5, 0.2
print('P(A union B):', p_a + p_b - p_intersection)
print('Independent intersection example:', 0.6 * 0.2)
```

## 3. Conditional probability

```python
p_intersection, p_b = 0.12, 0.3
print('P(A given B):', p_intersection / p_b)
```

## 4. Bayes' theorem and base rates

```python
prevalence = 0.02
sensitivity = 0.95
false_positive_rate = 0.05
p_positive = sensitivity * prevalence + false_positive_rate * (1 - prevalence)
p_condition_given_positive = sensitivity * prevalence / p_positive
print('P(positive):', p_positive)
print('P(condition given positive):', p_condition_given_positive)
```

## 5. Expected value and variance

```python
values = [0, 1, 2]
probabilities = [0.25, 0.5, 0.25]
mean = sum(x * p for x, p in zip(values, probabilities))
variance = sum((x - mean) ** 2 * p for x, p in zip(values, probabilities))
print('Expected value:', mean)
print('Variance:', variance)
```

## 6. Simulation

```python
import random
rng = random.Random(42)
trials = 10_000
heads = sum(rng.choice(['H', 'T']) == 'H' for _ in range(trials))
print('Estimated P(heads):', heads / trials)
```
