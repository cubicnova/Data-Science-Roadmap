# Conditional Statements in Python

Run each cell and change values to explore different branches.

## 1. `if` and `if...else`

```python
temperature = 34
if temperature > 30:
    print('Hot')

number = 13
if number % 2 == 0:
    print('Even')
else:
    print('Odd')
```

## 2. Multiple branches with `elif`

```python
score = 84
if not 0 <= score <= 100:
    grade = 'Invalid score'
elif score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'
print('Grade:', grade)
```

## 3. Combining conditions

```python
age = 21
has_id = True
is_suspended = False
if age >= 18 and has_id and not is_suspended:
    print('Entry allowed')
else:
    print('Entry denied')

day = 'Sunday'
if day in ('Saturday', 'Sunday'):
    print('Weekend')
else:
    print('Not a weekend day')
```

## 4. Nested conditions and safe checks

```python
items = ['admin', 'reports']
if items:
    if items[0] == 'admin':
        print('Admin list')
    else:
        print('First item is not admin')
else:
    print('List is empty')
```

## 5. Conditional expression and `None` check

```python
age = 17
status = 'Adult' if age >= 18 else 'Minor'
print(status)

value = None
if value is None:
    print('No value was supplied')
elif value == 0:
    print('The value is zero')
```

## 6. Classify data with conditions

```python
scores = [92, 71, 38, 84, 55]
for score in scores:
    if score >= 80:
        category = 'High'
    elif score >= 50:
        category = 'Medium'
    else:
        category = 'Low'
    print(score, category)
```
