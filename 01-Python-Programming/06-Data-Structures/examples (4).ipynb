# Data Structures in Python

Runnable examples of lists, tuples, sets, dictionaries, strings, mutability, and practical data processing.

## 1. Lists and slicing

```python
scores = [82, 91, 75, 91]
print('First:', scores[0])
print('Last:', scores[-1])
print('Slice:', scores[1:3])
scores.append(88)
scores[1] = 95
print('Updated:', scores)
```

## 2. Tuples and strings

```python
coordinates = (10.5, 20.2)
print('Coordinates:', coordinates)
single = (5,)
print('One-item tuple:', single, type(single))
text = 'Data Science'
print(text.lower())
print(text[0:4])
print(text[::-1])
```

## 3. Sets

```python
values = [1, 2, 2, 3, 3, 3]
print('Unique:', set(values))
a = {1, 2, 3}
b = {3, 4, 5}
print('Union:', a | b)
print('Intersection:', a & b)
print('Difference:', a - b)
```

## 4. Dictionaries

```python
student = {'name': 'Asha', 'score': 91}
student['score'] = 95
student['city'] = 'Indore'
print(student.get('email', 'Not provided'))
for key, value in student.items():
    print(key, value)
```

## 5. Mutability and copying

```python
original = [1, 2]
alias = original
alias.append(3)
print('Original after alias mutation:', original)
copy = original.copy()
copy.append(4)
print('Original after copy mutation:', original)
print('Copy:', copy)
```

## 6. Nested data and frequency counts

```python
students = [
    {'name': 'Asha', 'marks': [88, 91, 85]},
    {'name': 'Ravi', 'marks': [78, 82, 80]},
]
print(students[0]['name'], students[0]['marks'][1])
words = ['data', 'python', 'data', 'analysis']
frequencies = {}
for word in words:
    frequencies[word] = frequencies.get(word, 0) + 1
print(frequencies)
```
