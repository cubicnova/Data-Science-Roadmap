# Loops in Python

Runnable examples for `for`, `while`, `range`, loop control, nested loops, iteration helpers, and data processing.

## 1. `for` loops and `range()`

```python
for number in range(1, 6):
    print(number)

for character in 'data':
    print(character)

print(list(range(10, 0, -2)))
```

## 2. `while` loops

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

## 3. `break` and `continue`

```python
print('Break example:')
for number in range(1, 8):
    if number == 5:
        break
    print(number)

print('Continue example:')
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

## 4. Counters and accumulators

```python
numbers = [3, 7, 2, 8]
total = 0
even_count = 0
for number in numbers:
    total += number
    if number % 2 == 0:
        even_count += 1
print('Total:', total)
print('Even count:', even_count)
```

## 5. Nested loops

```python
for row in range(1, 4):
    for column in range(1, 3):
        print((row, column))

for row in range(1, 5):
    print('*' * row)
```

## 6. `enumerate()`, `zip()`, and loop `else`

```python
names = ['Asha', 'Ravi', 'Mina']
scores = [91, 85, 94]
for index, name in enumerate(names, start=1):
    print(index, name)
for name, score in zip(names, scores):
    print(name, score)

target = 7
for value in [2, 4, 6, 8]:
    if value == target:
        print('Found')
        break
else:
    print('Not found')
```

## 7. Data processing with loops

```python
scores = [82, 47, 35, 91, 68]
passing = []
for score in scores:
    if score >= 40:
        passing.append(score)
print('Passing:', passing)

values = [2, 3, 2, 5, 3, 2]
frequencies = {}
for value in values:
    if value in frequencies:
        frequencies[value] += 1
    else:
        frequencies[value] = 1
print('Frequencies:', frequencies)
```
