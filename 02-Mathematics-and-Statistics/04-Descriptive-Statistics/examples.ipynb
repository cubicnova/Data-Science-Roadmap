# Descriptive Statistics for Data Science

This notebook demonstrates summary statistics, missing-value inspection, category proportions, and potential outlier checks with pandas and NumPy.

## 1. Central tendency

```python
import numpy as np
import pandas as pd

values = [2, 3, 3, 4, 18]
s = pd.Series(values)
print('mean:', s.mean())
print('median:', s.median())
print('mode(s):', s.mode().tolist())
```

## 2. Spread and summary

```python
print('range:', s.max() - s.min())
print('sample variance:', s.var())
print('sample std:', s.std())
print(s.describe())
```

## 3. Quartiles and IQR

```python
q1, q3 = s.quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print('Q1:', q1, 'Q3:', q3, 'IQR:', iqr)
print('fences:', lower, upper)
print('potential outliers:', s[(s < lower) | (s > upper)].tolist())
```

## 4. Categorical frequency

```python
labels = pd.Series(['red', 'blue', 'red', 'green', 'blue', 'red'])
print(labels.value_counts())
print(labels.value_counts(normalize=True))
```

## 5. Missing values

```python
df = pd.DataFrame({'score': [80, None, 92, 75], 'group': ['A', 'A', 'B', None]})
print(df.isna().sum())
```

## 6. Correlation

```python
df_corr = pd.DataFrame({'hours': [1, 2, 3, 4, 5], 'score': [50, 55, 65, 70, 80]})
print(df_corr.corr(numeric_only=True))
```
