# Pandas Indexing and Filtering

Runnable examples for selecting, filtering, and safely updating DataFrames.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'name':['Asha','Dev','Mira','Ravi'],'age':[21,24,20,27],'score':[78,91,85,62],'city':['Indore','Pune','Indore','Bhopal']})
df
```

## 1. Column selection

```python
print(df['score'])
print(df[['name', 'city']])
```

## 2. Label-based and position-based selection

```python
print('loc label:', df.loc[2, 'name'])
print('iloc position:', df.iloc[2, 1])
print('first two rows:\n', df.iloc[:2])
```

## 3. Boolean filters

```python
print(df[df['score'] >= 80])
print(df[(df['age'] >= 21) & (df['city'] == 'Indore')])
print(df[df['city'].isin(['Indore', 'Pune'])])
```

## 4. Range and missing-value filters

```python
df_missing = df.copy()
df_missing.loc[1, 'score'] = np.nan
print('age range:\n', df[df['age'].between(20, 25)])
print('missing scores:\n', df_missing[df_missing['score'].isna()])
```

## 5. query and where

```python
print(df.query('age > 22 and score > 70'))
print(df['score'].where(df['score'] >= 80))
```

## 6. Index management

```python
by_name = df.set_index('name')
print(by_name.loc['Mira'])
print(by_name.reset_index())
print('index unique:', df.index.is_unique)
```

## 7. Safe updates and copies

```python
updated = df.copy()
updated.loc[updated['score'] < 70, 'score'] = 70
high_scores = updated.loc[updated['score'] >= 80].copy()
print(updated)
print('high scores:\n', high_scores)
```

## 8. String filtering and sorting

```python
print(df[df['name'].str.startswith('M')])
print(df[df['city'].str.contains('ind', case=False, na=False)])
print(df.sort_values('score', ascending=False))
```

## 9. Mini challenge: employee filtering

```python
employees = pd.DataFrame({'name':['Asha','Dev','Mira'],'department':['IT','HR','IT'],'salary':[70000,65000,55000]})
result = employees.loc[(employees['department'] == 'IT') & (employees['salary'] > 60000), ['name', 'salary']]
print(result)
assert result['name'].tolist() == ['Asha']
```
