# NumPy Statistical Functions

A hands-on notebook covering descriptive statistics, percentiles, axes, missing values, and correlation.

```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)
```

## 1. Mean and median

```python
scores = np.array([72, 85, 90, 68, 95, 80])
print('mean:', np.mean(scores))
print('median:', np.median(scores))
```

## 2. Minimum, maximum, range, and indices

```python
a = np.array([8, 3, 12, 6, 9])
print('min:', np.min(a), 'max:', np.max(a), 'range:', np.ptp(a))
print('argmin:', np.argmin(a), 'argmax:', np.argmax(a))
```

## 3. Population and sample variation

```python
a = np.array([2, 4, 6, 8])
print('population variance:', np.var(a), 'population std:', np.std(a))
print('sample variance:', np.var(a, ddof=1), 'sample std:', np.std(a, ddof=1))
```

## 4. Percentiles and quantiles

```python
a = np.arange(1, 11)
print('quartiles:', np.percentile(a, [25, 50, 75]))
print('0.8 quantile:', np.quantile(a, 0.8))
```

## 5. Reduce along axes

```python
data = np.array([[10, 20, 30], [40, 50, 60]])
print('column means:', np.mean(data, axis=0))
print('row means:', np.mean(data, axis=1))
print('keepdims shape:', np.mean(data, axis=0, keepdims=True).shape)
```

## 6. Weighted average

```python
values = np.array([70, 90])
weights = np.array([1, 3])
print(np.average(values, weights=weights))
```

## 7. Missing values

```python
a = np.array([1.0, 2.0, np.nan, 4.0])
print('ordinary mean:', np.mean(a))
print('NaN-aware mean:', np.nanmean(a))
print('NaN-aware median:', np.nanmedian(a))
```

## 8. Correlation and covariance

```python
x = np.array([1, 2, 3])
y = np.array([2, 4, 6])
print('correlation matrix:\n', np.corrcoef(x, y))
print('covariance matrix:\n', np.cov(x, y))
```

## 9. Boolean counts

```python
values = np.array([2, 5, 8, 11, 14])
print('values greater than 7:', np.sum(values > 7))
```

## 10. Outlier impact

```python
a = np.array([20, 21, 22, 23, 200])
print('mean:', np.mean(a), 'median:', np.median(a))
```
