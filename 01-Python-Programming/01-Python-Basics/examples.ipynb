# Python Variables and Data Types

This notebook demonstrates variables, built-in data types, conversions, strings, collections, mutability, and a small Data Science example.

**Instructions:** Run cells in order, change values, and observe the outputs.

## 1. Variables and Dynamic Typing

```python
name = "Rahul"
age = 21
height = 1.75
is_student = True

print(name, age, height, is_student)

value = 100
print(type(value).__name__)
value = "Python"
print(type(value).__name__)
```

## 2. Built-in Data Types

```python
values = [10, 3.14, 2 + 3j, "Python", True, [1, 2], (1, 2), {1, 2}, {"name": "Amit"}, None]

for item in values:
    print(repr(item), type(item).__name__)
```

## 3. Numeric Operations and Formatting

```python
price = 250.0
quantity = 4
total = price * quantity
print(f"Total cost: {total:.2f}")

first = 20
second = 5
print("Sum:", first + second)
print("Difference:", first - second)
print("Product:", first * second)
print("Quotient:", first / second)
```

## 4. Strings and f-strings

```python
first_name = "Priya"
last_name = "Sharma"
full_name = f"{first_name} {last_name}"
print(full_name)
print(full_name.upper())
print(full_name.lower())
print("Length:", len(full_name))
```

## 5. Collections

```python
numbers = [10, 20, 30]
coordinates = (10, 20)
languages = {"Python", "SQL", "Python"}
student = {"name": "Amit", "marks": 85}

numbers.append(40)
student["marks"] = 90
print("List:", numbers)
print("Tuple:", coordinates)
print("Set:", languages)
print("Dictionary:", student)
```

## 6. Type Conversion and Safe Conversion

```python
age = int("21")
price = float("99.50")
label = str(100)
print("Age next year:", age + 1)
print("Price:", price)
print("Label:", label)

def convert_to_float(value):
    try:
        return float(value)
    except ValueError:
        return None

print(convert_to_float("12.5"))
print(convert_to_float("Python"))
```

## 7. Mutability and References

```python
first = [1, 2, 3]
second = first
second.append(4)

print("First:", first)
print("Second:", second)
print("Same object:", first is second)

copy_of_first = first.copy()
copy_of_first.append(5)
print("Original after shallow copy is changed:", first)
print("Copy:", copy_of_first)
```

## 8. Multiple Assignment and Unpacking

```python
x, y, z = 10, 20, 30
print(x, y, z)
x, z = z, x
print("After swap:", x, y, z)
```

## 9. Mini Data Science Example

```python
student = {
    "name": "Neha",
    "marks": [80, 90, 85],
    "is_enrolled": True,
}
marks = student["marks"]
average = sum(marks) / len(marks)

print(f"Student: {student['name']}")
print(f"Number of subjects: {len(marks)}")
print(f"Total marks: {sum(marks)}")
print(f"Average marks: {average:.2f}")
```
