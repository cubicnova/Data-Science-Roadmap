# Matplotlib: Examples and Practice

A hands-on introduction to creating and customizing charts. Run cells in order.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

print('Matplotlib, NumPy, and Pandas are ready.')
```

## 1. Create a dataset

```python
df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "revenue": [120, 135, 128, 160, 172, 185],
    "orders": [12, 14, 13, 16, 18, 20],
    "ad_spend": [10, 12, 11, 15, 17, 19]
})
df
```

## 2. A basic line plot

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(df["month"], df["revenue"], marker="o", label="Revenue")
ax.set(title="Monthly Revenue", xlabel="Month", ylabel="Revenue (units)")
ax.legend()
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
plt.show()
```

## 3. Bar chart and horizontal bar chart

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(df["month"], df["orders"])
ax.set(title="Orders by Month", xlabel="Month", ylabel="Orders (count)")
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(df["month"], df["revenue"])
ax.set(title="Revenue by Month", xlabel="Revenue (units)", ylabel="Month")
fig.tight_layout()
plt.show()
```

## 4. Scatter plot

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(df["ad_spend"], df["revenue"], alpha=0.75)
ax.set(title="Advertising Spend vs Revenue", xlabel="Advertising spend", ylabel="Revenue")
fig.tight_layout()
plt.show()
```

## 5. Histogram and box plot

```python
order_values = [18, 20, 21, 22, 22, 24, 25, 25, 27, 28, 30, 31, 35, 38, 42]
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(order_values, bins=6, edgecolor="black")
ax.set(title="Order Value Distribution", xlabel="Order value", ylabel="Frequency")
plt.show()
```

```python
weekday = [42, 45, 47, 48, 50, 52, 55, 57]
weekend = [35, 39, 44, 49, 51, 58, 62, 70]
fig, ax = plt.subplots(figsize=(7, 4))
ax.boxplot([weekday, weekend], tick_labels=["Weekday", "Weekend"])
ax.set(title="Order Values by Day Type", ylabel="Order value")
plt.show()
```

## 6. Styling, legend, and reference line

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(df["month"], df["revenue"], marker="o", linestyle="--", linewidth=2, label="Revenue")
ax.axhline(150, linestyle=":", label="Target = 150")
ax.set(title="Revenue Against Target", xlabel="Month", ylabel="Revenue")
ax.legend()
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
plt.show()
```

## 7. Annotate a point

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(df["month"], df["revenue"], marker="o")
peak_index = df["revenue"].idxmax()
peak_month = df.loc[peak_index, "month"]
peak_value = df.loc[peak_index, "revenue"]
ax.annotate(f"Peak: {peak_value}",
            xy=(peak_index, peak_value),
            xytext=(peak_index - 1.5, peak_value - 15),
            arrowprops={"arrowstyle": "->"})
ax.set(title="Annotated Revenue Trend", xlabel="Month", ylabel="Revenue")
plt.show()
```

## 8. Multiple subplots

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8), layout="constrained")
axes[0, 0].plot(df["month"], df["revenue"], marker="o")
axes[0, 0].set_title("Revenue Trend")
axes[0, 1].bar(df["month"], df["orders"])
axes[0, 1].set_title("Orders")
axes[1, 0].hist(order_values, bins=6, edgecolor="black")
axes[1, 0].set_title("Order Value Distribution")
axes[1, 1].scatter(df["ad_spend"], df["revenue"])
axes[1, 1].set_title("Ad Spend vs Revenue")
fig.suptitle("Sales Performance Overview")
plt.show()
```

## 9. Axis limits and log scales

```python
x = np.arange(1, 6)
y = [10, 100, 1000, 10000, 100000]
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, y, marker="o")
ax.set_yscale("log")
ax.set(title="Values Across Orders of Magnitude", xlabel="Observation", ylabel="Value (log scale)")
ax.grid(True, which="both", alpha=0.25)
plt.show()
```
