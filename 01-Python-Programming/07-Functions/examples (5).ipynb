# 07 – Functions in Python

Runnable examples of function definitions, arguments, return values, variable-length arguments, lambdas, higher-order functions, recursion, and data-cleaning helpers.

```python
def greet(name="Learner"):
    """Return a friendly greeting."""
    return f"Hello, {name}!"

print(greet())
print(greet("Asha"))
```

## Parameters, return values, and defaults

```python
def calculate_total(price, quantity=1, discount=0):
    """Calculate a total after applying a discount fraction."""
    return price * quantity * (1 - discount)

print(calculate_total(100, 3, 0.10))
print(calculate_total(price=50, quantity=2))
```

## Variable-length arguments

```python
def sum_all(*numbers):
    return sum(numbers)

def build_profile(**details):
    return details

print(sum_all(2, 4, 6))
print(build_profile(name="Mina", role="data analyst"))
```

## Docstrings and type hints

```python
def celsius_to_fahrenheit(celsius: float) -> float:
    """Convert Celsius to Fahrenheit."""
    return celsius * 9 / 5 + 32

print(celsius_to_fahrenheit(0))
print(celsius_to_fahrenheit.__doc__)
```

## Lambda functions and sorting

```python
students = [("Ravi", 78), ("Mina", 92), ("Dev", 85)]
ordered = sorted(students, key=lambda student: student[1], reverse=True)
print(ordered)
triple = lambda value: value * 3
print(triple(7))
```

## Higher-order functions and closures

```python
def apply_twice(value, operation):
    return operation(operation(value))

def make_multiplier(factor):
    def multiply(value):
        return value * factor
    return multiply

print(apply_twice(-3, abs))
double = make_multiplier(2)
print(double(11))
```

## Recursion and validation

```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

## Data-science utility functions

```python
def clean_names(names):
    """Trim whitespace and omit empty names."""
    return [name.strip() for name in names if name.strip()]

def summary(values):
    if not values:
        raise ValueError("values cannot be empty")
    return {"count": len(values), "total": sum(values), "mean": sum(values) / len(values)}

print(clean_names([" Asha ", "", " Ravi", "   "]))
print(summary([10, 20, 30, 40]))
```
