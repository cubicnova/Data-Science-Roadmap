# Visualization Fundamentals

This notebook demonstrates foundational chart types and practical design choices using small example datasets. Run from top to bottom.

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Pandas:', pd.__version__)
print('Matplotlib ready')
```

## 1. Create a small dataset

We will use a simple sales dataset to practice grouping and charting.

```python
sales = pd.DataFrame({
    "month": ["Jan", "Jan", "Feb", "Feb", "Mar", "Mar", "Apr", "Apr"],
    "product": ["Books", "Games", "Books", "Games", "Books", "Games", "Books", "Games"],
    "revenue": [120, 90, 140, 110, 135, 125, 160, 130],
    "orders": [12, 9, 14, 11, 13, 12, 16, 13]
})
sales
```

## 2. Bar chart: compare categories

```python
product_revenue = sales.groupby("product")["revenue"].sum().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(product_revenue.index, product_revenue.values)
ax.set(title="Total Revenue by Product", xlabel="Product", ylabel="Revenue (units)")
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
plt.show()
```

## 3. Line chart: show change over time

```python
monthly_revenue = sales.groupby("month", sort=False)["revenue"].sum()
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(monthly_revenue.index, monthly_revenue.values, marker="o")
ax.set(title="Monthly Revenue", xlabel="Month", ylabel="Revenue (units)")
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
plt.show()
```

## 4. Histogram: inspect a distribution

```python
order_values = [18, 20, 21, 22, 22, 24, 25, 25, 27, 28, 30, 31, 35, 38, 42]
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(order_values, bins=6, edgecolor="black")
ax.set(title="Distribution of Order Values", xlabel="Order value", ylabel="Frequency")
fig.tight_layout()
plt.show()
```

## 5. Scatter plot: explore a relationship

A positive association does not prove that one variable causes the other.

```python
ad_spend = [2, 3, 4, 5, 6, 7, 8, 9]
revenue = [18, 20, 25, 27, 31, 35, 36, 42]
fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(ad_spend, revenue, alpha=0.75)
ax.set(title="Advertising Spend and Revenue", xlabel="Advertising spend (units)", ylabel="Revenue (units)")
fig.tight_layout()
plt.show()
```

## 6. Box plot: compare distributions

```python
weekday = [42, 45, 47, 48, 50, 52, 55, 57]
weekend = [35, 39, 44, 49, 51, 58, 62, 70]
fig, ax = plt.subplots(figsize=(7, 4))
ax.boxplot([weekday, weekend], tick_labels=["Weekday", "Weekend"])
ax.set(title="Order Values by Day Type", ylabel="Order value")
fig.tight_layout()
plt.show()
```

## 7. Customize labels and layout

Titles should describe the question; axes should identify values and units.

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(["North", "South", "East", "West"], [42, 35, 51, 46])
ax.set_title("Orders by Region")
ax.set_xlabel("Region")
ax.set_ylabel("Orders (count)")
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
plt.show()
```

## 8. Pandas quick plotting

Pandas plotting is convenient for exploration; Matplotlib provides more detailed control.

```python
monthly_orders = sales.groupby("month", sort=False)["orders"].sum()
ax = monthly_orders.plot(kind="bar", figsize=(7, 4), legend=False)
ax.set(title="Orders by Month", xlabel="Month", ylabel="Orders (count)")
plt.tight_layout()
plt.show()
```
