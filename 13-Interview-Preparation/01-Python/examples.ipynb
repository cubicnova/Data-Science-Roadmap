## 1. Built-in collections and mutability

Lists are mutable, tuples are immutable containers, sets hold unique hashable items, and dictionaries map hashable keys to values.

```python
numbers = [1, 2, 3]
numbers.append(4)
coordinates = (10, 20)
tags = {"python", "data", "python"}
scores = {"Asha": 91, "Ravi": 84}

print(numbers)
print(coordinates)
print(tags)
print(scores["Asha"])
```

**Interview prompt:** Why can a tuple contain a list even though tuples are immutable?

```python
record = ([1, 2], "active")
record[0].append(3)
print(record)
# record[0] = [9]  # TypeError: tuple item assignment is not supported
```

## 2. Equality, identity, and copying

```python
first = [1, 2]
alias = first
shallow = first.copy()

print(first == alias)   # same value
print(first is alias)   # same object
print(first == shallow) # same value
print(first is shallow) # different outer list
```

```python
import copy

nested = [[1, 2], [3, 4]]
shallow = copy.copy(nested)
deep = copy.deepcopy(nested)
nested[0].append(99)

print(shallow)  # inner list is shared
print(deep)     # nested lists were copied
```

## 3. Strings, slicing, and comprehensions

```python
text = "Data Science"
print(text.lower())
print(text[0:4])
print(text[::-1])

squares = [n * n for n in range(6) if n % 2 == 0]
print(squares)
```

## 4. Dictionaries, sets, and frequency counting

```python
from collections import Counter

items = ["ml", "python", "ml", "sql", "python", "ml"]
counts = Counter(items)
print(counts)
print("sql" in set(items))

left = {"python", "sql"}
right = {"sql", "statistics"}
print(left | right)  # union
print(left & right)  # intersection
```

## 5. Functions, argument passing, and safe defaults

```python
def describe(name, *skills, level="junior", **metadata):
    return {
        "name": name,
        "skills": skills,
        "level": level,
        "metadata": metadata,
    }

print(describe("Asha", "Python", "SQL", level="intermediate", city="Indore"))
```

```python
def append_value(value, items=None):
    if items is None:
        items = []
    items.append(value)
    return items

print(append_value(1))
print(append_value(2))  # fresh list
```

## 6. Sorting and key functions

```python
people = [
    {"name": "Ravi", "score": 82},
    {"name": "Asha", "score": 95},
    {"name": "Meera", "score": 95},
]

ranked = sorted(people, key=lambda person: (-person["score"], person["name"]))
print(ranked)
```

## 7. Scope and closures

```python
def make_multiplier(factor):
    def multiply(value):
        return value * factor
    return multiply

double = make_multiplier(2)
print(double(7))
```

## 8. Exceptions and context managers

```python
def parse_integer(value):
    try:
        number = int(value)
    except (TypeError, ValueError) as error:
        print(f"Invalid integer: {error}")
        return None
    else:
        return number

print(parse_integer("42"))
print(parse_integer("four"))
```

Use `with` to ensure resources are cleaned up even when work raises an exception. This example uses an in-memory stream so it does not create files.

```python
from io import StringIO

with StringIO("name,score\nAsha,95\n") as stream:
    content = stream.read()
print(content)
```

## 9. Iterators and generators

```python
def running_total(values):
    total = 0
    for value in values:
        total += value
        yield total

for total in running_total([2, 3, 5]):
    print(total)
```

**Interview prompt:** How does `yield` differ from `return`, and why can a generator help with large inputs?

## 10. Object-oriented Python

```python
class ModelResult:
    category = "classification"

    def __init__(self, model_name, accuracy):
        self.model_name = model_name
        self.accuracy = accuracy

    def summary(self):
        return f"{self.model_name}: accuracy={self.accuracy:.2f}"

    @classmethod
    def default(cls):
        return cls("baseline", 0.0)

    @staticmethod
    def is_valid_accuracy(value):
        return isinstance(value, (int, float)) and not isinstance(value, bool) and 0 <= value <= 1

result = ModelResult("logistic-regression", 0.91)
print(result.summary())
print(ModelResult.default().summary())
print(ModelResult.is_valid_accuracy(0.91))
```

## 11. Coding patterns

```python
def two_sum(numbers, target):
    seen = {}
    for index, number in enumerate(numbers):
        complement = target - number
        if complement in seen:
            return [seen[complement], index]
        seen[number] = index
    raise ValueError("No pair sums to target")

assert two_sum([2, 7, 11, 15], 9) == [0, 1]
print("Two Sum test passed")
```

```python
def moving_average(values, window):
    if window <= 0:
        raise ValueError("window must be positive")
    if window > len(values):
        return []
    total = sum(values[:window])
    result = [total / window]
    for end in range(window, len(values)):
        total += values[end] - values[end - window]
        result.append(total / window)
    return result

assert moving_average([1, 2, 3, 4], 2) == [1.5, 2.5, 3.5]
print("Moving average test passed")
```
