# NumPy Reshaping and Joining

Hands-on examples for reshaping arrays, changing dimensions, joining, and splitting. Run the cells in order.

```python
import numpy as np
np.set_printoptions(suppress=True)
```

## 1. Inspect shape

```python
a = np.arange(12)
print('array:', a)
print('shape:', a.shape, '| ndim:', a.ndim, '| size:', a.size)
```

## 2. Reshape and infer dimensions

```python
a = np.arange(12)
print(a.reshape(3, 4))
print('inferred shape:', np.arange(18).reshape(3, -1).shape)
```

## 3. Flattening

```python
matrix = np.arange(6).reshape(2, 3)
print('flatten:', matrix.flatten())
print('ravel:', matrix.ravel())
print('shares memory:', np.shares_memory(matrix, matrix.ravel()))
```

## 4. Add and remove dimensions

```python
v = np.array([10, 20, 30])
print('row:', v[np.newaxis, :].shape)
print('column:', v[:, np.newaxis].shape)
x = np.zeros((1, 3, 1))
print('squeezed:', np.squeeze(x).shape)
```

## 5. Concatenate along an existing axis

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
print(np.concatenate((a, b), axis=0))
```

## 6. Concatenate columns

```python
a = np.array([[1], [2], [3]])
b = np.array([[10, 20], [30, 40], [50, 60]])
print(np.concatenate((a, b), axis=1))
```

## 7. Stack creates a new axis

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print('axis 0:\n', np.stack((a, b), axis=0))
print('axis 1:\n', np.stack((a, b), axis=1))
```

## 8. Convenience join functions

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print('vstack:\n', np.vstack((a, b)))
print('hstack:', np.hstack((a, b)))
print('dstack shape:', np.dstack((a, b)).shape)
```

## 9. Split evenly and unevenly

```python
a = np.arange(12)
print('equal:', np.split(a, 3))
print('uneven:', np.array_split(np.arange(10), 3))
```

## 10. Split a matrix by rows and columns

```python
m = np.arange(24).reshape(4, 6)
print('row section shapes:', [x.shape for x in np.vsplit(m, 2)])
print('column section shapes:', [x.shape for x in np.hsplit(m, 3)])
```

## 11. Round-trip split and join

```python
a = np.arange(12)
parts = np.array_split(a, 4)
recombined = np.concatenate(parts)
print('recombined:', recombined)
assert np.array_equal(a, recombined)
```

## 12. Shape validation

```python
a = np.zeros((2, 3))
b = np.ones((2, 2))
joined = np.concatenate((a, b), axis=1)
print('joined shape:', joined.shape)
assert joined.shape == (2, 5)
```

## 13. Mini challenge: batches

```python
data = np.arange(24)
batches = data.reshape(4, 6)
first_half, second_half = np.split(batches, 2, axis=0)
print('batches:', batches.shape)
print('first half:', first_half.shape, '\n', first_half)
print('second half:', second_half.shape, '\n', second_half)
assert first_half.shape == second_half.shape == (2, 6)
```
