# Operators in Python

This notebook demonstrates arithmetic, comparison, assignment, logical, bitwise, membership, identity, precedence, and data-science filtering operations.

## 1. Arithmetic operators

`/` performs true division, `//` floor division, `%` returns a remainder, and `**` raises a value to a power.

```python
a, b = 17, 5
print('addition:', a + b)
print('subtraction:', a - b)
print('multiplication:', a * b)
print('true division:', a / b)
print('floor division:', a // b)
print('remainder:', a % b)
print('power:', 3 ** 4)
```

## 2. Assignment operators

Augmented assignment updates a variable using an operation.

```python
balance = 100
balance += 25
balance -= 10
balance *= 2
balance //= 5
print('updated balance:', balance)
```

## 3. Comparison and logical operators

Comparisons return Booleans. Logical operators combine conditions.

```python
score = 76
age = 20
has_id = True
print('passed:', score >= 40)
print('valid score:', 0 <= score <= 100)
print('entry allowed:', age >= 18 and has_id)
print('weekend:', 'Saturday' == 'Saturday' or 'Saturday' == 'Sunday')
print('empty username:', not '')
```

## 4. Membership and identity

Use `in` to test membership. Use `==` for value equality and `is` for identity checks, commonly with `None`.

```python
tools = ['Python', 'Pandas', 'NumPy']
print('Pandas available:', 'Pandas' in tools)
record = {'name': 'Asha', 'score': 90}
print('score is a key:', 'score' in record)
print('90 is a key:', 90 in record)
first = [1, 2]
second = [1, 2]
print('same values:', first == second)
print('same object:', first is second)
result = None
print('missing result:', result is None)
```

## 5. Bitwise operators

Bitwise operations work on integer bits. The following example models permission flags.

```python
print('6 in binary:', bin(6))
print('6 & 3:', 6 & 3)
print('6 | 3:', 6 | 3)
print('6 ^ 3:', 6 ^ 3)
print('3 << 2:', 3 << 2)
print('16 >> 2:', 16 >> 2)
READ, WRITE, EXECUTE = 1, 2, 4
permissions = READ | WRITE
print('write allowed:', (permissions & WRITE) != 0)
print('execute allowed:', (permissions & EXECUTE) != 0)
```

## 6. Precedence and parentheses

Multiplication happens before addition. Exponentiation associates from the right.

```python
print('2 + 3 * 4 =', 2 + 3 * 4)
print('(2 + 3) * 4 =', (2 + 3) * 4)
print('2 ** 3 ** 2 =', 2 ** 3 ** 2)
```

## 7. Operator applications in data science

Comparisons and logical conditions can filter records. Python list comprehensions are a useful introduction to the filtering patterns used in analytical workflows.

```python
scores = [82, 47, 35, 91, 68]
passing = [score for score in scores if score >= 40]
high_scores = [score for score in scores if score >= 70]
print('passing scores:', passing)
print('high scores:', high_scores)

def even_percentage(numbers):
    if not numbers:
        return 0.0
    even_count = sum(number % 2 == 0 for number in numbers)
    return even_count / len(numbers) * 100

print('even percentage:', even_percentage([1, 2, 3, 4]))
print('empty input percentage:', even_percentage([]))
```
