# 11 – Object-Oriented Programming in Python

Examples of classes, objects, methods, properties, inheritance, polymorphism, composition, and dataclasses.

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def describe(self):
        return f"{self.name}: {self.score}"

student = Student("Asha", 92)
print(student.name)
print(student.describe())
```

## Class attributes and class methods

```python
class Employee:
    company = "Example Ltd"
    def __init__(self, name):
        self.name = name
    @classmethod
    def company_name(cls):
        return cls.company
    @staticmethod
    def is_valid_name(name):
        return bool(name.strip())

employee = Employee("Ravi")
print(employee.company, employee.name)
print(Employee.company_name())
print(Employee.is_valid_name("Mina"))
```

## Properties and validation

```python
class Product:
    def __init__(self, price):
        self.price = price
    @property
    def price(self):
        return self._price
    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("price cannot be negative")
        self._price = value

product = Product(25.5)
print(product.price)
try:
    product.price = -1
except ValueError as exc:
    print(type(exc).__name__ + ":", exc)
```

## Inheritance and polymorphism

```python
class Animal:
    def speak(self):
        return "Some sound"
class Cat(Animal):
    def speak(self):
        return "Meow"
class Bird(Animal):
    def speak(self):
        return "Tweet"

def make_it_speak(animal):
    return animal.speak()

print(make_it_speak(Cat()))
print(make_it_speak(Bird()))
```

## Special methods

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
    def __str__(self):
        return f"{self.title} ({self.pages} pages)"
    def __len__(self):
        return self.pages

book = Book("Python Basics", 240)
print(book)
print(len(book))
```

## Composition

```python
class Engine:
    def start(self):
        return "Engine started"
class Car:
    def __init__(self, engine):
        self.engine = engine
    def start(self):
        return self.engine.start()

car = Car(Engine())
print(car.start())
```

## Dataclasses

```python
from dataclasses import dataclass, field

@dataclass
class DatasetInfo:
    name: str
    columns: list[str] = field(default_factory=list)

info = DatasetInfo("sales", ["date", "amount"])
print(info)
```
