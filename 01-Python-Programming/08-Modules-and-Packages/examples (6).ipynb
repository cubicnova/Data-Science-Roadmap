## Selected imports and aliases

````python
from statistics import mean, median
import random as rnd
print(mean([3, 6, 9]))
print(median([8, 2, 5, 1, 9]))
print(1 <= rnd.randint(1, 10) <= 10)
````

## Paths and dates

````python
from pathlib import Path
from datetime import date
print(Path.cwd())
print(Path('README.md').name)
print(date.today().isoformat())
````

## JSON encoding and decoding

````python
import json
record = {"name": "Asha", "score": 95, "active": True}
encoded = json.dumps(record)
decoded = json.loads(encoded)
print(encoded)
print(decoded["name"], decoded["score"])
````

## Collections utilities

````python
from collections import Counter, defaultdict
print(Counter("banana"))
visits = defaultdict(int)
visits["home"] += 1
visits["home"] += 1
print(dict(visits))
````

## Import metadata and main guard

````python
import math
print("Module name:", math.__name__)

def main():
    print("This demo runs when main() is called.")

if __name__ == "__main__":
    main()
````

## Simulating a small reusable module

````python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

def multiply(a, b):
    """Return the product of two numbers."""
    return a * b

print(add(3, 4))
print(multiply(3, 4))
print(add.__doc__)
````
