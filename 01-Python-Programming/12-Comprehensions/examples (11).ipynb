# Python Comprehensions

This notebook demonstrates list, set, and dictionary comprehensions, filters, conditional expressions, nested comprehensions, and generator expressions. Run the cells from top to bottom.

## 1. List comprehensions

```python
squares = [n ** 2 for n in range(1, 6)]
print(squares)
```

```python
evens = [n for n in range(10) if n % 2 == 0]
print(evens)
```

## 2. Filtering versus conditional transformation

```python
filtered = [n for n in range(6) if n % 2 == 0]
labels = ['even' if n % 2 == 0 else 'odd' for n in range(6)]
print('filtered:', filtered)
print('labels:', labels)
```

## 3. Transforming strings

```python
words = ['  ada ', 'grace', 'linus']
cleaned = [word.strip().title() for word in words]
print(cleaned)
```

## 4. Set comprehensions

```python
words = ['cat', 'dog', 'python', 'owl']
unique_lengths = {len(word) for word in words}
print(sorted(unique_lengths))
```

## 5. Dictionary comprehensions

```python
square_map = {n: n ** 2 for n in range(1, 6)}
print(square_map)
```

```python
prices = {'pen': 10, 'book': 80, 'bag': 120}
with_tax = {item: round(price * 1.10, 2) for item, price in prices.items()}
print(with_tax)
```

## 6. Nested comprehensions

```python
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [value for row in matrix for value in row]
print(flat)
```

```python
pairs = [(x, y) for x in [1, 2] for y in ['a', 'b']]
print(pairs)
```

```python
matrix = [[1, 2, 3], [4, 5, 6]]
transposed = [[row[col] for row in matrix] for col in range(len(matrix[0]))]
print(transposed)
```

## 7. Generator expressions

```python
total = sum(n ** 2 for n in range(1, 101))
print(total)
```

## 8. A practical mini-example

```python
records = [
    {'name': 'Asha', 'score': 82},
    {'name': 'Ben', 'score': 67},
    {'name': 'Chen', 'score': 91},
]
passing = [record['name'] for record in records if record['score'] >= 70]
score_by_name = {record['name']: record['score'] for record in records}
print('passing:', passing)
print('score map:', score_by_name)
```
