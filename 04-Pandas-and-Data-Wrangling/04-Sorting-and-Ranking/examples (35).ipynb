# Sorting and Ranking in Pandas

This notebook demonstrates sorting values and indexes, null placement, ranking ties, group ranking, top-N selection, and ordered categories.

```python
import pandas as pd

pd.set_option('display.max_columns', None)
df = pd.DataFrame({
    'name': ['Asha', 'Ravi', 'Meera', 'Kabir', 'Zoya', 'Ishan'],
    'department': ['IT', 'Sales', 'IT', 'Sales', 'HR', 'IT'],
    'score': [88, 74, 95, 74, 91, 88],
    'experience': [3, 5, 2, 7, 4, 1],
    'city': ['Indore', 'Pune', 'Indore', 'Delhi', 'Pune', 'Delhi']
})
df
```

## 1. Sort one column

```python
df.sort_values('score')
```

```python
df.sort_values('score', ascending=False)
```

## 2. Sort by multiple columns

```python
df.sort_values(['department', 'score'], ascending=[True, False])
```

## 3. Case-insensitive sorting

```python
names = pd.DataFrame({'name': ['zoya', 'Asha', 'meera', 'RAVI']})
names.sort_values('name', key=lambda s: s.str.lower())
```

## 4. Missing values and index sorting

```python
missing = pd.DataFrame({'item': ['A', 'B', 'C'], 'score': [80, None, 95]})
print('Missing first:')
display(missing.sort_values('score', na_position='first'))
print('Missing last:')
display(missing.sort_values('score', na_position='last'))
```

```python
indexed = df.set_index('name')
print('Rows by index:')
display(indexed.sort_index())
print('Columns by label:')
display(df.sort_index(axis=1))
```

## 5. Ranking and ties

```python
scores = pd.Series([90, 80, 90, 70], index=['Asha', 'Ravi', 'Meera', 'Kabir'])
print('Average:', scores.rank(ascending=False).to_dict())
print('Min:', scores.rank(ascending=False, method='min').to_dict())
print('Dense:', scores.rank(ascending=False, method='dense').to_dict())
```

## 6. Rank within each department

```python
ranked = df.copy()
ranked['department_rank'] = ranked.groupby('department')['score'].rank(ascending=False, method='dense')
ranked.sort_values(['department', 'department_rank', 'name'])
```

## 7. Top-N and bottom-N

```python
print('Top three scores:')
display(df.nlargest(3, 'score'))
print('Two least experienced:')
display(df.nsmallest(2, 'experience'))
```

## 8. Ordered categories

```python
priority_order = pd.CategoricalDtype(categories=['Low', 'Medium', 'High'], ordered=True)
priorities = pd.DataFrame({'task': ['T1', 'T2', 'T3'], 'priority': pd.Series(['High', 'Low', 'Medium'], dtype=priority_order)})
priorities.sort_values('priority')
```

## 9. Sort and reset index

```python
df.sort_values(['score', 'experience'], ascending=[False, True], ignore_index=True)
```

## Key takeaways

- Use `sort_values()` for data values and `sort_index()` for labels.
- Use `rank(ascending=False)` when the highest value should rank first.
- Select a tie method deliberately, and use `groupby().rank()` for within-group ranks.
- Sort before using `head()` when you mean top or bottom records.
