# NumPy Practice Exercises

A notebook of runnable examples spanning the core NumPy topics.

```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)
```

## 1. Array creation and metadata

```python
a = np.arange(1, 13).reshape(3, 4)
print(a)
print('shape:', a.shape, 'ndim:', a.ndim, 'size:', a.size, 'dtype:', a.dtype)
```

## 2. Indexing and slicing

```python
m = np.arange(25).reshape(5, 5)
print('central block:\n', m[1:4, 1:4])
print('last column:', m[:, -1])
print('reversed:', np.arange(5)[::-1])
```

## 3. Boolean filtering and `where`

```python
a = np.arange(1, 31)
print('multiples of 3:', a[a % 3 == 0])
temps = np.array([28, 32, 35, 27, 31])
print('hot labels:', np.where(temps > 30, 1, 0))
```

## 4. Vectorized arithmetic

```python
a = np.array([8, 12, 16])
b = np.array([2, 4, 8])
print('sum:', a + b, 'product:', a * b)
print('squares:', np.arange(1, 10).reshape(3, 3) ** 2)
```

## 5. Aggregation and axes

```python
m = np.arange(12).reshape(3, 4)
print('row sums:', np.sum(m, axis=1))
print('column means:', np.mean(m, axis=0))
print('overall median:', np.median(m))
```

## 6. Broadcasting

```python
m = np.array([[1, 2, 3], [4, 5, 6]])
v = np.array([10, 20, 30])
print(m + v)
```

## 7. Reshaping and joining

```python
a = np.arange(24).reshape(2, 3, 4)
print('3-D shape:', a.shape)
x = np.concatenate((np.array([[1, 2]]), np.array([[3, 4]])), axis=0)
print(x)
print('stack shape:', np.stack((np.arange(4), np.arange(4, 8))).shape)
```

## 8. Splitting

```python
parts = np.split(np.arange(18), 3)
print(parts)
assert np.array_equal(np.concatenate(parts), np.arange(18))
```

## 9. Statistical summaries and NaNs

```python
a = np.array([12., 15., 18., 21., 24., 30.])
print('mean:', np.mean(a), 'median:', np.median(a), 'std:', np.std(a), 'p75:', np.percentile(a, 75))
print('NaN-aware mean:', np.nanmean([2., 4., np.nan, 8.]))
```

## 10. Reproducible random numbers

```python
rng1 = np.random.default_rng(42)
rng2 = np.random.default_rng(42)
a = rng1.integers(0, 100, size=10)
b = rng2.integers(0, 100, size=10)
print(a)
print('reproducible:', np.array_equal(a, b))
assert np.array_equal(a, b)
```

## 11. Mini-project: marks analysis

```python
marks = np.array([[80, 70, 90], [75, 85, 88], [90, 92, 84], [65, 78, 80], [88, 82, 95]])
print('subject means:', np.mean(marks, axis=0))
print('highest per student:', np.max(marks, axis=1))
print('overall median:', np.median(marks))
```

## 12. Mini-project: sales analysis

```python
sales = np.array([[10, 20, 15], [12, 18, 17], [14, 22, 16], [9, 25, 20], [16, 21, 18], [18, 19, 22], [13, 24, 19]])
daily_total = np.sum(sales, axis=1)
print('product totals:', np.sum(sales, axis=0))
print('best day index:', np.argmax(daily_total))
print('lowest day index:', np.argmin(daily_total))
```
